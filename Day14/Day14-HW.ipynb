**Q1**
Algorithms are able to analyze huge amounts of data faster and more accurately than humans which can sometimes be a good thing. They don’t get tired or emotional so they can do work more consistently. they can actually reduce some human bias when designed carefully (like favoritism or inconsistency) and make decisions based on data instead of feeling. They also can help us spot patterns we’d never notice ourselves, like in healthcare or fraud detection.
The problem is that algorithms learn from our bias data. So even if the algorithm seems okay, it can end up repeating or even worsening existing inequalities. one example is facial recognition misidentifying people with darker skin because the technology was trained with primarily white people. Another issue is transparency; it can be difficult to know how an algorithm came to its decision, which makes it tough to fix. sometimes we rely on them too much.

**Q2**

What could be influencing this result? I also think it’s important to test for fairness and make sure decisions aren’t only based on accuracy but on impact too.

**Q3**

2  This article gives a good overview of the current state of research and policy around bias and fairness in AI (fair-AI). It describes how bias arises from data, algorithms, and the organizational context. It then surveys methods and resources for detecting, mitigating and accounting for bias, and links these to policy and legal layers via the “NoBIAS” project. The authors highlight that although many technical fairness metrics exist, the challenge spans legal, ethical, social and organizational dimensions.

3  This really emphasised how much of algorithmic bias isn’t just a “data problem” but a total system problem with algorithms, people, and policy. It made me rethink how in our own data-science work I might be tempted to focus only on accuracy or predictive power and neglect fairness, transparency or accountability. I agree with the authors that we need think about and frameworks that embed fairness from the start rather than add it on at the end. What stood out to me was the emphasis on embedding a bias management layer and a legal/regulatory layer. it reminds me that in my future data-science major or projects I should always consider the context and not just the code.
